### 문제1. 3의 배수 출력

- **문제 설명**  
1부터 n까지의 숫자 중에서 3의 배수만 출력하는 함수를 작성하세요.

- **함수 설명**  
`print_multiples_of_three(n: int) -> int`:  
  - n: 3의 배수를 구하고자 하는 범위의 최대값입니다.

- **입출력 예시**   

    - 입력1:

    ```
    7
    ```

    - 출력1:

    ```
    3
    6
    ```

    - 입력2:
    
    ```
    30
    ```

    - 출력2:

    ```
    3
    6
    9
    12
    15
    18
    21
    24
    27
    30
    ```

````python
# 문제
def print_multiples_of_three(n: int) -> None:
  '''
  1부터 n까지의 숫자 중에서 3의 배수만 출력하는 함수

  Args:
    n (int): 3의 배수를 구하고자 하는 범위의 최대값
  '''
  try:
    parsed_n = int(n)
    for i in range(3, parsed_n + 1, 3):
      print(i)
  except:
    print("숫자를 입력해주세요.")
````

### 문제2. 모음 제거하기

- **문제 설명**  
문자열에서 모음(a, e, i, o, u)을 제거한 문자열을 반환하는 함수를 작성하세요.

- **함수 설명**  
`remove_vowels(s: string) -> string`:  
  - s: 영어 소문자 알파벳으로 이루어진 문자열입니다.

- **입출력 예시**   

    - 입력1:

    ```
    "codeit"
    ```

    - 출력1:

    ```
    "cdt"
    ```

    - 입력2:
    
    ```
    "python programming"
    ```

    - 출력2:

    ```
    "pythn prgrmmng"
    ```



````python
# 문제
def remove_vowels(s : str) -> str:
  '''
  문자열에서 모음(a, e, i, o, u)을 제거한 문자열을 반환하는 함수

  Args:
    s (str): 모음을 제거하고자 하는 문자열

  Returns:
    str: 모음이 제거된 문자열 또는 문자열이 잘못된 경우 오류 메시지
  '''
  try:
    arg_s = str(s)
    return "".join(c for c in arg_s if c not in ['a', 'e', 'i', 'o', 'u'])
  except:
    return "문자열이 잘못되었습니다"
````

### 문제3. 수 크기 비교하기

- **문제 설명**  
두 개의 숫자를 입력받아, 큰 숫자와 작은 숫자를 각각 반환하는 함수를 작성하세요.
(단, 동일한 숫자를 2회 입력하는 경우에는 순서 없이 한 번만 출력되도록 해 주세요.)

- **함수 설명**

  `compare_numbers(num1, num2)`:  
  - num1, num2 : 정수형

- **입출력 예시**

  - 입력 1:
    ```
    10, 20
    ```

  - 출력 1:
    ```
    (20, 10)
    ```

  - 입력 2:
    ```
    30, 30
    ```

  - 출력 2:
    ```
    30
    ```

````python
# 문제
def compare_numbers(num1: int, num2: int) -> tuple[int, int] | int | str:
  '''
  두 개의 숫자를 입력받아, 큰 숫자와 작은 숫자를 각각 반환하는 함수
  단, 숫자가 같을 경우 하나의 숫자만 반환합니다.

  Args:
    num1 (int) : 첫번째 숫자
    num2 (int) : 두번째 숫자

  Returns:
    tuple[int, int] | int | str:
      - tuple[int, int]: num1이 num2보다 크면 (큰 숫자, 작은 숫자)
      - int: num1이 num2보다 작거나 같으면 num1을 반환
      - str: 숫자가 아닐 경우 오류 메시지
  '''
  try:
    parsed_num1 = int(num1)
    parsed_num2 = int(num2)
    if parsed_num1 == parsed_num2:
      return parsed_num1
    else:
      return max(parsed_num1, parsed_num2), min(parsed_num1, parsed_num2)
  except:
    return "숫자를 입력해주세요."
````

````python
# 프로그램 실행
print(compare_numbers(10, 20))
````

- **입출력 예시**

    - 입력1:

    ```

    ```

    - 출력2:

````python
# 문제
def sum_even_numbers(num: int) -> None:
  '''
  사용자가 입력한 숫자 n까지의 짝수 합을 구하는 함수

  Args:
    num (int): 짝수 합을 구하고자 하는 범위의 최대값
  '''
  try:
    parsed_num = int(num)
    if parsed_num == 0: # parsed_num이 0 이면 종료 출력 후 종료
      print("종료")
      return
    else:
      sum = 0
      for i in range(1, parsed_num+1):
        if i % 2 == 0:
          sum += i
      print(f"1부터 {parsed_num}까지의 짝수 합은 {sum}입니다.")
  except:
    print("숫자를 입력해주세요.")
````

- **함수 설명**

  `divide_numbers(a, b)`
  - 사용자에게 두개의 숫자를 차례대로 입력 받습니다.
  - 첫번째 숫자를 두번째 숫자로 나눕니다.
  - 결과값을 반환합니다. 단, 예외가 발생하면 "잘못된 입력입니다."를 반환합니다.
  - `try-except` 구문을 활용하여 함수를 작성해보세요. 이 문법에 대해서 구글링을 해도 좋습니다.

- **입출력 예시**

- 입력1:
    ```
    42, 7
    ```

- 출력1:
    ```
    6
    ```

- 입력2:
    ```
    135, 0
    ```

- 출력2:
    ```
    "잘못된 입력입니다."
    ```

````python
# 문제
def divide_numbers(a: int, b: int) -> float:
  '''사용자가 입력한 두 숫자를 나눕니다.
  예외가 발생하면 "잘못된 입력입니다."를 문자열로 출력합니다

  Args:
    a (int) : 첫번째 숫자
    b (int) : 두번째 숫자

  Returns:
    float : 입력된 첫번째 수에 두번째 수를 나눈 값을 반환합니다

  Exceptioins:
    입력값이 숫자가 아닐 경우 "잘못된 입력입니다" 문자열을 출력합니다
  '''
  try:
    num1 = int(a)
    num2 = int(b)
    if num2 == 0: # num2이 0 이면 나눌 수 없다
      print("잘못된 입력입니다.")
    else:
      return num1 / num2
  except:
    print("잘못된 입력입니다.")
````

### 문제1. 문자열 압축 게임



- **문제 설명**

  주어진 문자열에서 연속해서 같은 문자가 반복되는 경우, 그 문자와 반복 횟수를 이용해 새로운 형태로 문자열을 압축합니다. 예를 들어, "aabcccccaaa" 문자열은 "a2b1c5a3"로 압축될 수 있습니다.

- **함수 설명**  
`compress_string(s: str) -> str`:  
  - s: 압축하고자 하는 문자열입니다.

- **입출력 예시**

    - 입력1:

    ```
    "aaabbaaa"
    ```

    - 출력1:

    ```
    "a3b2a3"
    ```

    - 입력2:
    
    ```
    "aabcccccaaa"
    ```

    - 출력2:

    ```
    "a2b1c5a3"
    ```

````python
# 문제
def compress_string(s: str) -> None:
  '''
  주어진 문자열에서 연속해서 같은 문자가 반복되는 경우,
  그 문자와 반복 횟수를 이용해 새로운 형태로 문자열을 압축하여 출력합니다.

  Args:
    s (str) : 압축하고자 하는 문자열
  '''
  try:
    output = ""
    if s == None or len(s) == 0:
      print("입력값이 비어있습니다")
      return

    target_str = str(s)

    target_char = target_str[0] # 첫번째 문자를 할당
    target_count = 0
    i = 0

    while i < len(target_str):
      if target_char == target_str[i]: # 할당된 문자와 문자열의 대상 문자가 같다면
        target_count += 1 # 중복 카운트를 1 증가시킨다.
      else: # 할당된 문자와 문자열의 대상 문자가 다르다면
        output += target_char # 그동안 중복체크 했던 문자를 결과값에 저장한다
        output += str(target_count) # 이어서 중복 카운트를 문자 뒤에 저장한다
        target_char = target_str[i] # 대상 문자를 할당 문자로 설정한다
        target_count = 1 # 중복 카운트를 1로 한다.
      i += 1

    output += target_char # 문자열 끝까지 탐색한 다음 마지막 중복체크 문자를 결과값에 저장한다
    output += str(target_count) # 이어서 중복 카운트를 문자 뒤에 저장한다.

    print(output)
  except:
    print("잘못된 입력입니다.")
````

````python
# 프로그램 실행
compress_string("aaabbaaa")
````

### 문제2. 덧셈 프로그램



- **문제 설명**
이 프로그램은 두 개의 숫자를 입력받아 그 합을 계산합니다. 입력받은 값이 숫자 형식이 아니면, 사용자에게 숫자 형식의 입력을 요청합니다. 숫자 입력이 완료되면 두 숫자의 합을 출력합니다.

- **함수 설명**

  `add_numbers()`
  - 이 함수는 `input()`을 사용하여 사용자로부터 두 개의 숫자를 입력받습니다.
  - 입력된 값이 숫자가 아닐 경우, "올바른 숫자를 입력하세요."라는 메시지를 출력하고 다시 입력을 요청합니다.
  - 두 숫자의 합을 계산하고 결과를 출력합니다.

- **입출력 예시**

    - 입력:
        ```
        첫 번째 숫자를 입력하세요: 10
        첫 번째 숫자를 입력하세요: twenty
        첫 번째 숫자를 입력하세요: 10
        두 번째 숫자를 입력하세요: 20
        ```

    - 출력:
        ```
        올바른 숫자를 입력하세요.
        10 + 20 = 30
        ```

````python
# 문제
def input_numbers() -> tuple[int, int]:
  '''
  사용자로부터 숫자를 입력받고, 그것을 정수 쌍으로 변환합니다.
  변환이 불가할 경우 None을 반환합니다.

  Returns:
    tuple[int, int]: 개별적으로 입력된 정수 숫자 2개를 튜플로 반환합니다

  Exceptioins:
    입력이 숫자가 아닐 경우 None으로 채워진 튜플을 반환합니다.
  '''
  try:
    input_num = input("첫 번째 숫자를 입력하세요: ")
    num1 = int(input_num)
    input_num = input("두 번째 숫자를 입력하세요: ")
    num2 = int(input_num)
    return num1, num2
  except:
    return None, None

def add_numbers() -> None:
  '''
  두 개의 숫자를 입력받아 그 합을 계산하여 산술식과 함께 출력합니다

  Exceptioins:
    입력이 숫자가 아닐 경우 첫 번째 숫자부터 다시 입력받습니다
  '''
  while True:
    num1, num2 = input_numbers()
    if num1 == None or num2 == None:
      print("올바른 숫자를 입력하세요.")
      continue
    else:
      print(f"{num1} + {num2} = {num1 + num2}")
      break
````

````python
# 프로그램 실행
add_numbers()
````

### 문제3. 카운트다운 타이머

- **문제 설명**  
  사용자로부터 시간(초)을 입력받아 해당 시간만큼 카운트 다운을 진행한 후, "타이머 종료!"를 출력하고 프로그램이 종료됩니다. 매 초마다 남은 시간을 화면에 출력합니다.

- **함수 설명**     
  
  `count_down(seconds: int)`:  
  - seconds: 카운트 다운할 시간(초)입니다.
  - 입력받은 초만큼 카운트 다운을 진행하며, 매 초마다 남은 시간을 출력합니다.
  - 시간이 종료되면 "타이머 종료!"를 출력하고 프로그램을 종료합니다.

- **입출력 예시**

  - 입력:
    ```
    타이머를 시작할 시간(초)을 입력하세요: 5
    ```

  - 출력:
    ```
    타이머 시작!
    5
    4
    3
    2
    1
    타이머 종료!
    ```

````python
# 문제
import asyncio

async def countdown_timer(seconds : int) -> None:
  '''
  사용자로부터 시간(초)을 입력받아 해당 시간만큼 카운트 다운을 진행한 후,
  "타이머 종료!"를 출력하고 프로그램이 종료됩니다. 매 초마다 남은 시간을 화면에 출력합니다.

  Args:
    seconds (int) : 카운트 다운할 시간(초)

  Exceptioins:
    입력이 숫자가 아닐 경우 오류 메시지를 출력합니다.
  '''
  try:
    seconds = int(seconds)
  except:
    print("잘못된 입력입니다.")
    return

  print("타이머 시작!")
  while seconds > 0:
    print(f"{seconds}")
    await asyncio.sleep(1)
    seconds -= 1
  print("타이머 종료!")
````

### 문제1. 시간 추적 클래스(TimeTracker)

- **실습 설명**

  시간을 관리하고 추적하는 `TimeTracker` 클래스를 구현하는 프로젝트를 시작합니다. 시간 관리 기능은 특히 프로젝트 작업, 운동, 공부 시간 등 다양한 활동의 지속 시간을 측정하는 데 유용합니다.

  `TimeTracker` 클래스는 다음 기능을 제공해야 합니다:

  1. **시작 시간 설정**: 사용자가 활동을 시작할 때의 시간을 기록합니다.
  2. **종료 시간 설정**: 사용자가 활동을 종료할 때의 시간을 기록합니다.
  3. **경과 시간 계산**: 활동의 시작과 종료 사이의 시간 차이를 계산합니다.

  이 클래스의 인스턴스를 사용하여 각각의 활동에 대해 별도의 시간 추적을 할 수 있어야 합니다.

- **구현해야 할 메소드**

  - `start`: 현재 시간을 시작 시간으로 설정합니다.
  - `stop`: 현재 시간을 종료 시간으로 설정하고 경과 시간을 계산합니다.
  - `get_elapsed_time`: 마지막으로 기록된 시작 시간과 종료 시간 사이의 경과 시간을 분 단위로 반환합니다.

- **실습 결과 예시**

  다음과 같은 코드를 실행했을 때의 출력 예시입니다:

  ```python
  study_session = TimeTracker()
  study_session.start()
  # 1시간 30분 경과(가정)
  study_session.stop()

  print("활동 시간:", study_session.get_elapsed_time(), "분")
  ```

  예상 출력:

  ```
  공부한 시간: 90 분
  ```

- **요구 사항**

  1. 실제 시간을 추적하려면 Python의 `datetime` 모듈을 사용하여 현재 시간을 `datetime.now()`로 가져올 수 있습니다.
  2. 경과 시간은 분 단위로 반환해야 합니다.

````python
# 문제
from datetime import datetime

class TimeTracker:
  '''
  공부 시간을 관리하고 추적하여 문자열로 출력하는 클래스.

  Methods:
    start(): 사용자가 활동을 시작할 때의 시간을 기록합니다.
    stop(): 사용자가 활동을 종료할 때의 시간을 기록합니다.
    get_elapsed_time(): 활동의 시작과 종료 사이의 시간 차이를 계산하여 분 단위로 출력합니다.
  '''
  def __init__(self):
    self.start_time = None
    self.end_time = None

  def start(self) -> None:
    self.start_time = datetime.now()

  def stop(self) -> None:
    self.end_time = datetime.now()

  def get_elapsed_time(self) -> int:
    return int((self.end_time - self.start_time).total_seconds() // 60)
````

````python
# 사용 예시
# 활동 시작
study_session = TimeTracker()
study_session.start()
````

````python
# 활동 종료
study_session.stop()
print("공부한 시간:", study_session.get_elapsed_time(), "분")
````

  - 전화번호(phone number)

  클래스는 이 정보를 효율적으로 관리할 수 있는 기능을 제공해야 합니다.


- **구현해야 할 메소드**

  - `__init__`: 객체를 생성할 때 이름, 전화번호, 이메일 주소를 초기화합니다.
  - `__str__`: 연락처의 정보를 예쁘게 출력할 수 있는 문자열로 반환합니다. 이 문자열은 연락처 정보를 한눈에 알아볼 수 있도록 포맷팅됩니다.

  ```
  전화번호: 010-1234-5678
  이메일: jane@example.com
  ```

- **요구 사항**

  - 모든 입력 데이터는 문자열로 처리해야 합니다.
  - 연락처 정보를 적절하게 포맷팅하여 출력할 수 있어야 합니다.

````python
# 문제
class Contact:
  '''
  주소록 관리 시스템 클래스.

  Methods:
    __init__(self, name: str, phone_number: str, email: str):
      객체를 생성할 때 이름(name), 전화번호(phone number), 이메일 주소(email address) 를 저장합니다.
    __str__(self):
      연락처의 정보를 예쁘게 출력할 수 있는 문자열로 반환합니다.
  '''
  def __init__(self, name: str, phone_number: str, email: str):
    try:
      self.name = str(name)
      self.phone_number = str(phone_number)
      self.email = str(email)
    except ValueError as e:
      print(f"잘못된 입력입니다: {e}")

  def __str__(self):
    return f"이름 : {self.name}\n전화번호 : {self.phone_number}\n이메일 : {self.email}"
````

### 문제1. 투표 시스템 클래스

- **실습 설명**

  간단한 투표 시스템을 위한 `VoteSystem` 클래스를 구현하는 프로젝트를 시작합니다. 이 시스템은 후보자 목록을 관리하고, 각 후보자에 대한 투표를 집계하는 기능을 제공합니다.

  `VoteSystem` 클래스는 다음 기능을 제공해야 합니다:

  - 후보자 등록
  - 투표 기능
  - 투표 결과 조회

- **구현해야 할 메소드**

  - `add_candidate`: 후보자를 등록합니다. 후보자의 이름을 입력받아 목록에 추가합니다.
  - `vote`: 특정 후보자에게 투표합니다. 투표하려는 후보자의 이름을 입력받습니다.
  - `get_results`: 각 후보자의 투표 수를 출력합니다.

- **실습 결과 예시**

  - 다음과 같은 코드를 실행했을 때의 출력 예시입니다:

    ```python
    voting_system = VoteSystem()
    voting_system.add_candidate("Alice")
    voting_system.add_candidate("Bob")
    voting_system.add_candidate("Charlie")

    voting_system.vote("Alice")
    voting_system.vote("Alice")
    voting_system.vote("Bob")

    voting_system.get_results()
    ```

  - 예상 출력:

    ```
    Alice 후보가 성공적으로 등록되었습니다.
    Bob 후보가 성공적으로 등록되었습니다.
    Charlie 후보가 성공적으로 등록되었습니다.
    Alice에게 투표하였습니다.
    Alice에게 투표하였습니다.
    Bob에게 투표하였습니다.
    투표 결과:
    Alice: 2 votes
    Bob: 1 vote
    Charlie: 0 votes
    ```

- **요구 사항**

  - 후보자는 중복 등록될 수 없습니다.
  - 등록되지 않은 후보자에게 투표할 수 없습니다.
  - 각 후보자의 이름과 투표 수는 사전(dictionary)을 사용하여 관리합니다.

````python
# 문제
class VoteSystem:
  '''
  투표 시스템 클래스.

  Methods:
    add_candidate(name: str) -> None: 후보자 등록
    vote(name: str) -> None: 특정 후보자에게 투표
    get_results() -> None: 각 후보자의 투표 수를 출력
  '''
  candidates = {}

  def add_candidate(self, name: str) -> None:
    if name in self.candidates:
      print("이미 등록된 후보자입니다.")
      return
    self.candidates[name] = 0
    print(f"{name} 후보가 성공적으로 등록되었습니다.")

  def vote(self, name: str) -> None:
    if name in self.candidates:
      self.candidates[name] += 1
      print(f"{name}에게 투표하였습니다.")
    else:
      print("등록되지 않은 후보자입니다.")

  def get_results(self) -> None:
    output = "투표 결과:"
    for name, count in self.candidates.items():
      output += "\n"
      output += f"{name} : {count} votes"
    print(output)
````

````python
# 사용 예시
voting_system = VoteSystem()
voting_system.add_candidate("Alice")
voting_system.add_candidate("Bob")
voting_system.add_candidate("Charlie")

voting_system.vote("Alice")
voting_system.vote("Alice")
voting_system.vote("Bob")

voting_system.get_results()
````

  - `deposit`: 계좌에 금액을 입금합니다. 입금할 금액을 인자로 받고, 잔액을 업데이트합니다.


  - 예상 출력:

    ```
    김철수님의 계좌 123-456-789가 개설되었습니다. 초기 잔액: 100000원
    50000원이 입금되었습니다. 현재 잔액: 150000원
    20000원이 출금되었습니다. 현재 잔액: 130000원
    현재 잔액: 130000원
    ```


- **요구 사항**

  - 계좌에서 출금 시도 시 잔액보다 많은 금액을 출금하려고 하면, 출금을 거부하고 경고 메시지를 출력해야 합니다.
  - 모든 금액은 정수 또는 실수로 처리될 수 있어야 하며, 화폐 단위로만 입력받습니다.
  - 계좌 생성, 입금, 출금 및 잔액 조회 기능을 모두 구현해야 합니다.

````python
# 문제
class BankAccount:
  '''다음 정보와 기능을 제공합니다:

  Methods:
    __init__(self, account_number: str, account_holder: str, balance: float):
      객체 생성 시 계좌 번호, 소유자 이름, 초기 잔액을 설정합니다.
    deposit(self, amount: float) -> None: 계좌에 금액을 입금합니다.
    withdraw(self, amount: float) -> None: 계좌에서 금액을 출금합니다.
    get_balance(self) -> str: 현재 계좌 잔액을 반환합니다.
  '''
  def __init__(self, account_number: str, account_holder: str, balance: float):
    self.account_number = account_number
    self.account_holder = account_holder
    self.balance = balance
    print(f"{self.account_holder}님의 계좌 {self.account_number}가 개설되었습니다. 초기 잔액: {self.get_balance()}원")

  def deposit(self, amount: int) -> None:
    try:
      amount = int(amount)
    except:
      print("잘못된 입력입니다.")
      return
    self.balance += amount
    print(f"{amount}원이 입금되었습니다. 현재 잔액: {self.get_balance()}원")

  def withdraw(self, amount: int) -> None:
    try:
      amount = int(amount)
    except:
      print("잘못된 입력입니다.")
      return
    if self.balance < amount:
      print("잔액이 부족합니다.")
    else:
      self.balance -= amount
      print(f"{amount}원이 출금되었습니다. 현재 잔액: {self.get_balance()}원")

  def get_balance(self) -> str:
    '''balance가 float 타입이며 is_integer 검사를 통해 정수값에 가깝다면
    int로 형변환하고 그렇치 않다면 실수 자체를 문자열 반환합니다'''
    return str(int(self.balance) if isinstance(self.balance, float) and self.balance.is_integer() else self.balance)
````

````python
# 사용 예시
my_account = BankAccount("123-456-789", "김철수", 100000)
my_account.deposit(50000)
my_account.withdraw(20000)
print(f"현재 잔액: {my_account.get_balance()}원")

my_account.deposit(1234.5)
print(f"현재 잔액: {my_account.get_balance()}원")
my_account.withdraw(1234.5)
print(f"현재 잔액: {my_account.get_balance()}원")
````

````python
# 문제
class EmployeeManager:
  '''
  직원들의 정보를 저장하고, 전체 직원의 평균 급여를 계산하는 기능을 제공하는 클래스.

  Methods:
    __init__(self, name: str, salary: int):
      새로운 직원의 정보를 시스템에 추가합니다.
      직원의 이름과 급여 정보를 저장합니다.
    calculate_average_salary() -> None:
      클래스 메서드를 사용하여 모든 직원의 급여 평균을 계산합니다.
      이 메서드는 저장된 모든 직원의 급여 정보를 집계하여
      평균 급여를 계산하고 출력합니다.
  '''
  employees:dict[str, int] = {}

  def __init__ (self, name: str, salary: int):
    self.employees[name] = salary
    print(f"{name} 님이 추가되었습니다. 급여: {salary}원")

  @classmethod
  def calculate_average_salary(cls) -> None:
    if len(cls.employees) == 0:
      print("전체 직원의 평균 급여: 0")
      return

    total_salary = 0
    for salary in cls.employees.values():
      total_salary += salary
    print(f"전체 직원의 평균 급여: {total_salary // len(cls.employees)}원")
````

### 문제4. 프랜차이즈 레스토랑 관리 클래스

  - 모든 출력 메시지는 한국어로 제공되어야 합니다.
  - 각 메서드는 적절한 입력 검증과 예외 처리를 포함해야 합니다.
  - `sum_reservations` 클래스 메서드는 모든 지점에서의 예약 수를 효과적으로 합산하여 전체 예약 상태를 중앙에서 확인할 수 있게 합니다.

````python
# 문제
from datetime import datetime, date

DATETIME_FORMAT = "%Y-%m-%d"

class Reservation:
  '''
  고객의 예약 정보를 저장하고 정해진 문자열로 출력하는 클래스

  Methods:
    __init__(self, name: str, date: datetime, people: int):
      이름(name), 예약 일시(date), 인원 수(people)를 초기화합니다.
    __eq__(self, other): 예약 정보가 동일한지 반환합니다.
    __str__(self): 예약 정보를 문자열로 반환합니다.
  '''
  def __init__(self, name: str, date: datetime, people: int):
      self.name = name
      self.date = date
      self.people = people

  def __eq__(self, other):
    return self.name == other.name and self.date == other.date and self.people == other.people

  def __str__(self):
    return f"- {self.name}, {self.date.strftime(DATETIME_FORMAT)}, {self.people}명"

class ReservationSystem:
  '''
  각 레스토랑 지점의 예약을 관리하는 클래스.

  Methods:
    add_reservation(self, name: str, date_str: str, people: int) -> None:
      고객이 특정 지점, 예약 일시, 인원 수에 대한 예약을 요청하면 시스템에 추가합니다.
    cancel_reservation(self, name: str, date: str, people: int) -> None:
      고객이 예약을 취소할 수 있으며, 해당 예약을 시스템에서 제거합니다.
    list_reservations(self) -> None:
      특정 지점의 모든 예약 상황을 확인할 수 있습니다.
    sum_reservations(cls, reservation_system_list: list["ReservationSystem"]) -> int:
      모든 지점의 예약 수를 합산합니다.
      이 메서드는 모든 ReservationSystem 인스턴스의 예약 수를 합산하여 보여줍니다.
  '''
  def __init__(self, name):
    self.name = name
    self.reservations = []

  def add_reservation(self, name: str, date_str: str, people: int) -> None:
    try:
      date = datetime.strptime(date_str, DATETIME_FORMAT)
    except:
      print(f"날짜 형식이 잘못되었습니다 (년도-월-일 순으로 입력하세요): {date_str}")
      return
    target_reservation = Reservation(name, date, people)
    self.reservations.append(target_reservation)

  def cancel_reservation(self, name: str, date: str, people: int) -> None:
    for reservation in self.reservations:
      if reservation.name == name and reservation.date == date and reservation.people == people:
        self.reservations.remove(reservation)

  def list_reservations(self) -> None:
    output = f"{self.name} 예약 목록:\n"
    for reservation in self.reservations:
        output += str(reservation)
    print(output)

  @classmethod
  def sum_reservations(cls, reservation_system_list: list["ReservationSystem"]) -> int:
    total_reservations = 0
    for reservation_system in reservation_system_list:
        total_reservations += len(reservation_system.reservations)
    return total_reservations
````

````python
# 사용 예시
restaurant1 = ReservationSystem("강남점")
restaurant2 = ReservationSystem("홍대점")

restaurant1.add_reservation("홍길동", "2024-05-20", 4)
restaurant2.add_reservation("김철수", "2024-05-21", 2)

restaurant1.list_reservations()
restaurant2.list_reservations()

total_reservations = ReservationSystem.sum_reservations([restaurant1, restaurant2])
print(f"전체 레스토랑 예약 수: {total_reservations}")
````

````python
### 문제를 위한 파일 생성코드입니다. 문제를 풀기 전 이 코드를 반드시 실행해주세요.

import csv
# (시뮬레이션을 위한) 예시 파일 작성코드
def write_tasks_to_csv(filename, tasks):
    with open(filename, 'w', newline='') as csvfile:
        fieldnames = ['할일', '소요시간']
        writer = csv.DictWriter(csvfile, fieldnames=fieldnames)

        writer.writeheader()
        for task, time in tasks:
            writer.writerow({'할일': task, '소요시간': time})

# 할 일과 소요 시간 데이터
tasks = [
    ('공부하기', 120),
    ('운동하기', 60),
    ('가계부정리', 30),
    ('영화보기', 150)
]

# CSV 파일로 저장
write_tasks_to_csv('tasks.csv', tasks)
````

````python
import csv

# 문제
def load_tasks(filename: str) -> list[tuple[str, str]]:
  '''
  사용자로부터 CSV 파일 경로(filename)를 입력받아 파일을 읽습니다.
  파일은 각 할 일과 그에 소요되는 시간을 포함합니다.

  Args:
    filename (str): CSV 파일 경로

  Returns:
    list[tuple[str, str]]: csv 테이블에서 불려진 할일 목록
  '''
  with open(f'./{filename}', 'r', encoding='utf-8') as file_handle:
    reader = csv.reader(file_handle)
    tasks = list(reader)
    return tasks

def suggest_tasks(tasks: list[tuple[str, str]], time_limit: int, time_method: bool) -> None:
  '''
  남은 시간(time_limit)을 입력받습니다.
  이 시간을 기준으로 할 수 있는 최대한의 할 일을 수행하거나
  최대한 많은 시간을 쓰는 법을 선택하여 문자열로 출력합니다.

  Args:
    tasks (list[tuple[str, str]]): 할 일 목록
    time_limit (int): 남은 시간
    time_method (bool): 시간 소모 방법
      - True : Greedy 알고리즘 사용 - 최대한 많은 일을 하는 방법
      - False : Knapsack Problem 알고리즘 사용 - 최대한 많은 시간을 쓰는 방법
  '''
  if time_method:
    solve_greedy_max_time(tasks, time_limit)
  else:
    solve_knapsack_max_time(tasks, time_limit)

def solve_greedy_max_time(task_list: list[tuple[str, str]], time_limit: int) -> None:
  try:
    int(task_list[0][1])
  except:
    del task_list[0] # 제일 위쪽의 컬럼 필드를 삭제한다
  task_list.sort(key=lambda x: int(x[1])) # 소요 시간이 작은 순으로 정렬한다
  print("[Greedy] 선택된 할 일 목록:")
  i = 0
  while i < len(task_list):
    task = task_list[i]
    if int(task[1]) <= time_limit:
      print(f"{i + 1}. {task[0]} - 예상 소요 시간: {task[1]}분")
      time_limit -= int(task[1])
    i += 1

# 아래는 AI에게 물어서 얻은 Knapsack Problem (배낭 문제) 알고리즘
def solve_knapsack_max_time(task_list: list[tuple[str, str]], time_limit: int) -> None:
    n = len(task_list)
    # DP 테이블 초기화
    dp = [[0] * (time_limit + 1) for _ in range(n + 1)]

    # 1. DP 테이블 채우기
    for i in range(1, n + 1):
        name = task_list[i-1][0]
        try:
            duration = int(task_list[i-1][1]) # CSV 문자열을 숫자로 변환
        except ValueError:
            continue # 숫자가 아닌 데이터()가 섞여있다면 건너뜀

        value = duration

        for w in range(1, time_limit + 1):
            if duration <= w:
                dp[i][w] = max(dp[i-1][w], dp[i-1][w-duration] + value)
            else:
                dp[i][w] = dp[i-1][w]

    max_filled_time = dp[n][time_limit]

    # 2. 역추적 (Backtracking)으로 선택된 작업 찾기
    selected_tasks = []
    w = time_limit
    for i in range(n, 0, -1):
        if dp[i][w] != dp[i-1][w]:
            name = task_list[i-1][0]
            duration = int(task_list[i-1][1])
            selected_tasks.append((name, duration))
            w -= duration

    print("\n[Knapsack] 선택된 할 일 목록")
    if not selected_tasks:
        print("수행할 수 있는 일이 없습니다.")
    else:
        for i, (name, duration) in enumerate(selected_tasks):
            print(f"{i + 1}. {name} - 예상 소요 시간: {duration}분")

    print(f"총 사용 시간: {max_filled_time}분 / 남은 시간: {time_limit - max_filled_time}분")

def main():
    tasks = load_tasks('tasks.csv')
    time_limit = int(input("남은 시간(분): "))
    time_method_input = input("최대한 많은 일 하기[Y] / 최대한 많은 시간 쓰기[N]: ")
    time_method = True if time_method_input.upper() == 'Y' else False
    suggest_tasks(tasks, time_limit, time_method)
````

### 문제2. 수학 퍼즐: 소수의 합

- **실습 설명**

- **기능 설명**
  - 소수 판별: 주어진 수까지의 모든 자연수에 대해 소수 여부를 판별합니다. (힌트 : 에라토스테네스의 체)
  - 소수 합 계산: 판별된 소수들의 합을 계산합니다.
  - 결과 출력: 계산된 소수의 합을 출력합니다. 출력 형식은 사용자가 입력한 수까지의 소수 합과 그 값을 포함합니다.

- **입출력 예시**

  - 입력:

  ```
  어떤 수까지의 소수의 합을 구하시겠습니까?: 10
  ```
  - 출력:

  ```
  10 이하 소수의 합은 17입니다.
  ```

````python
# 문제
import math

def is_prime(num: int) -> bool:
  '''
  인자로 주어진 num이 2 미만이거나 짝수인 경우를 제외하고
  num의 제곱근을 구하여 num의 약수를 검출(있다면 합성수)
  단, 2는 소수이므로 True를 반환한다

  Args:
    num (int): 소수 여부를 확인할 자연수

  Returns:
    True: 소수(prime number)
    False: 합성수(composite number)
  '''
  if num < 2:
    return False
  if num == 2:
    return True
  if num % 2 == 0:
    return False

  limit = int(math.isqrt(num))
  for i in range(3, limit + 1, 2):
    if num % i == 0:
      return False
  return True

def sum_of_primes(n: int) -> int:
  '''인자로 주어진 n까지의 모든 자연수 중에 소수인 것들을
  is_prime 메서드로 확인하고 소수에 해당되면 모두 더해서 반환한다

  Args:
    n (int): 자신보다 작은 소수 합을 구하기 위한 자연수

  Returns:
    int: n 이하 소수의 합
  '''
  result = 0
  for i in range(2, n + 1):
    if is_prime(i):
      result += i
  return result

def main():
  n = int(input("어떤 수까지의 소수의 합을 구하시겠습니까?: "))
  result = sum_of_primes(n)
  print(f"{n} 이하 소수의 합은 {result}입니다.")
````

  - 시스템은 사용자의 행동에 따라 적절한 피드백을 제공해야 합니다. (예: 도서가 없을 때, 회원 정보가 없을 때)


````python
# 문제
import uuid
from datetime import datetime, timedelta

class Book:
  '''
  도서 정보(제목, 저자, 출판년도, ISBN)를 저장하는 클래스

  Methods:
    __init__(self, title: str, author: str, publication_year: int, isbn: str):
      제목(title), 저자(author), 출판년도(publication_year), ISBN(isbn)을 초기화합니다.
  '''
  def __init__(self, title: str, author: str, publication_year: int, isbn: str):
    self.title = title
    self.author = author
    self.publication_year = publication_year
    self.isbn = isbn

class Member:
  '''
  회원 정보(이름, 회원번호, 대여 중인 도서 목록)를 저장하는 클래스

  Methods:
    __init__(self, name: str):
      이름(name)을 초기화합니다. 또한 member_id를 중복되지 않게 자동 설정합니다.
    borrow_book(self, book: Book) -> None:
      대여 중인 도서 목록에 도서를 추가합니다.
    return_book(self, book: Book) -> None:
      대여 중인 도서 목록에서 도서를 제거합니다
  '''
  def __init__(self, name: str):
    self.name = name
    self.member_id = uuid.uuid4().hex # 동명 이인을 구분하기 위한 고유 값을 자동으로 발행한다
    self.borrowed_books = []

  def borrow_book(self, book: Book) -> None:
    self.borrowed_books.append(book)

  def return_book(self, book: Book) -> None:
    if book in self.borrowed_books:
      self.borrowed_books.remove(book)

class Rental:
  '''
  대여 정보(회원 ID, 도서 ISBN, 대여일, 반납일)를 저장하는 클래스

  Methods:
    __init__(self, member: Member, book: Book, days = 7):
      member와 book 객체를 받아와 대여 정보를 초기화합니다.
    _process_rental(self) -> None:
      대여 프로세스를 실행합니다.
    terminate_rental(self) -> Book:
      반납 프로세스를 실행합니다.
  '''
  def __init__(self, member: Member, book: Book, days = 7):
    '''ID 대신 객체 자체를 받아 처리하고, 대여 로직을 여기서 수행'''
    self.member = member
    self.book = book
    self.rental_date = datetime.now()
    self.return_date = self.rental_date + timedelta(days=days)

    # 중요! 대여 프로세스 실행 (Rental을 생성하는 동시에 회원의 대여 목록 업데이트)
    self._process_rental()

  def _process_rental(self) -> None:
    '''내부적으로 멤버의 대여 목록에 책을 추가'''
    self.member.borrow_book(self.book)

  def terminate_rental(self) -> Book:
    '''멤버의 대여 목록에서 책을 제거'''
    self.member.return_book(self.book)
    return self.book  # 반납된 책 반환

class LibraryManagement:
  '''
  도서, 회원, 대여 정보를 관리하는 메소드와 데이터 구조를 포함하는 클래스

  Methods:
    __init__(self):
      도서 목록, 회원 목록, 대여 목록을 초기화합니다.
    add_book(self, title: str, author: str, publication_year: int, isbn: str) -> None:
      도서를 추가합니다.
    add_member(self, name: str) -> None:
      회원을 추가합니다.
    get_book_from_isbn(self, book_isbn: str) -> Book:
      도서 ISDN으로 해당 도서 객체를 찾습니다
    get_member_from_name(self, member_name: str) -> Member:
      회원 이름으로 해당 회원 객체를 찾습니다
    rent_book(self, book_isbn: str, member_name: str) -> None:
      도서를 대여합니다
    return_book(self, book_isbn: str, member_name: str) -> None:
      도서를 반납합니다
    print_books(self) -> None:
      도서 목록을 출력합니다
    print_members(self) -> None:
      회원 목록을 출력합니다
  '''
  def __init__(self):
    self.books = []
    self.members = []
    self.rentals = []

  def add_book(self, title: str, author: str, publication_year: int, isbn: str) -> None:
    duplicated = self.get_book_from_isbn(isbn)
    if duplicated:
      print(f"'{title}' 도서는 이미 도서관 소장 목록에 포함되어 있습니다.")
      return
    book = Book(title, author, publication_year, isbn)
    self.books.append(book)

  def add_member(self, name: str) -> None:
    namesake = self.get_member_from_name(name)
    if namesake:
      answer = input(f"{name}님은 이미 등록된 회원입니다. 이름 뒤에 숫자를 붙이시겠습니까? [Y/n]:")
      if answer == 'Y' or answer == 'y':
        name += str(self.get_member_count_from_name(name))
      else:
        print("회원 등록을 취소합니다.")
        return
    member = Member(name)
    self.members.append(member)

  def get_book_from_isbn(self, book_isbn: str) -> Book:
    for book in self.books:
      if book.isbn == book_isbn:
        return book
    return None

  def get_member_from_name(self, member_name: str) -> Member:
    for member in self.members:
      if member.name == member_name:
        return member
    return None

  def get_member_count_from_name(self, member_name: str) -> int:
    count = 0
    for member in self.members:
      if member_name in member.name:
        count += 1
    return count

  def rent_book(self, book_isbn: str, member_name: str) -> None:
    target_book = self.get_book_from_isbn(book_isbn)
    target_member = self.get_member_from_name(member_name)

    if not target_book:
      print("해당 도서는 현재 도서관 소장 목록에 포함되어 있지 않습니다.")
      return

    if not target_member:
      print(f"{member_name}님은 회원으로 가입되지 않으셨습니다. 회원 가입부터 진행해주세요.")
      return

    # 이미 대여된 책에 대한 예외처리
    for rental in self.rentals:
      if rental.book == target_book:
        print(f"'{target_book.title}' 도서는 이미 대여 중입니다.")
        return

    # Rental 클래스를 인스턴스화하면서 대여 처리 진행
    rental = Rental(target_member, target_book)
    self.rentals.append(rental)

    print(f"'{member_name}' 회원님이 '{target_book.title}' 도서를 대여하였습니다.")

  def return_book(self, book_isbn: str, member_name: str) -> None:
    target_book = self.get_book_from_isbn(book_isbn)
    target_member = self.get_member_from_name(member_name)

    if not target_book or not target_member:
      print(f"{member_name}님은 회원으로 가입되지 않으셨습니다.")
      return

    target_rental = None
    for rental in self.rentals:
      if rental.book == target_book and rental.member == target_member:
        target_rental = rental
        break

    if target_rental:
      target_rental.terminate_rental()
      self.rentals.remove(target_rental) # 대여 기록 목록에서 삭제
      print(f"'{member_name}' 회원님이 '{target_book.title}' 도서를 반납하였습니다.")
    else:
      print(f"'{member_name}' 회원님은 '{target_book.title}' 도서를 대여하고 있지 않습니다.")

  def print_books(self) -> None:
    print("도서 목록:")
    for book in self.books:
      print(f"- {book.title} (저자: {book.author}, 출판년도: {book.publication_year})")

  def print_members(self) -> None:
    print("회원 목록:")
    for member in self.members:
      rental_books_titles = [book.title for book in member.borrowed_books]
      rental_books_str = ', '.join(rental_books_titles) if rental_books_titles else "없음"
      print(f"- {member.name} (대여 중인 도서: {rental_books_str})")
````
